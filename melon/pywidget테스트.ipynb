```python
import ipywidgets as widgets
from IPython.display import display

# 텍스트 박스 생성
text_input = widgets.Text(
    description='첫 번째 값:',
    placeholder='여기에 입력하세요'
)

# 버튼 생성
button = widgets.Button(description='제출')

# 버튼 클릭 시 동작할 함수 정의
def on_button_click(b):
    first_value = text_input.value
    print(f"입력한 첫 번째 값: {first_value}")
    
    # 두 번째 입력 받기 위한 텍스트 박스 생성
    second_text_input = widgets.Text(
        description='두 번째 값:',
        placeholder=f"'{first_value}'을(를) 기반으로 입력하세요"
    )
    
    # 두 번째 버튼 생성
    second_button = widgets.Button(description='제출')
    
    # 두 번째 버튼 클릭 시 동작할 함수 정의
    def on_second_button_click(b):
        second_value = second_text_input.value
        print(f"입력한 두 번째 값: {second_value}")

    # 두 번째 버튼 클릭 이벤트 연결
    second_button.on_click(on_second_button_click)

    # 두 번째 입력 위젯 표시
    display(second_text_input, second_button)

# 버튼 클릭 이벤트 연결
button.on_click(on_button_click)

# 첫 번째 입력 위젯 표시
display(text_input, button)
```
